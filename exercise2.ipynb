```python
#Program to find maximum among three numbers
a=int(input())
b=int(input())
c=int(input())
if a>b and a>c:
  print("a is max")
elif b>a and b>c:
  print("b is max")
else:
  print("c is max")
```

```python
#leap year checking
year=int(input())
if year%8==0 and year%100!=0 and year%400!=0:
  print("Entered year is a leap year")
else:
  print("Entered year is not a leap year")
```

```python
#Example cod eon match case
day=int(input())
match day:
  case 1:
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid input")
```
