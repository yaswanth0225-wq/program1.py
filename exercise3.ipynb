```python
#list operation
#1.concatenation
L1=[1,2,3,4]
L2=["hi",5,6,7]
print(L1+L2)
```

```python
#2.repetation
print(L2*2)
```

```python
#3.indexing
#i.forward indexing
print(L1[3])
#2.backward indexing
print(L2[-3])
```

```python
#4.slicing
print(L1[0:3])
print(L2[0:3])
print(L1[:])
L3=[1,2,3,4]
print(L3[0:5:2])
print((L3[-4:-1]))#reversing
print(L2[::-3])
```

```python
#5.membership
if 3 in L3:
  print("yes")
if 10 not in L3:
  print("yes")
```

```python
#iteration
for i in L3:
  print(i,end=" ")
```

```python
#built in functions for python
a=[1,2,3,4,5]
#1.len()
print(len(a))
#2.sum()
print(sum(a))
#3.min
print(min(a))
#4.reverse
print(list(reversed(a)))
#5.sort
print(sorted(a))
#6.max
print(max(a))
```

```python
#list methods
a=[1,2,3,4,5]
b=[5,3,2,6,7,9]
#1.append
a.append("santhosh")
print(a)
#2.extend
b.extend(a)
print(b)
#3.count
print(a.count(3))
#4.pop
b.pop()
print(b)
#5.index
print(b.index(3))
#6.remove
b.remove(9)
print(b)
#7.reverse
a.reverse()
print(a)
#8.sort
c=[20,56,34,2,1]
c.sort()
print(c)
```

```python
#program that prints sum of list elements without using sum()
a=[1,4,5,6,8]
sum=0
for i in a:
  sum+=i
print(sum)
```

```python
#print the list elements at odd index
list=[1,3,2,5,6,9,8,13]
for i in range(1,len(list),2):
  print(list[i],end=" ")
```

```python
#print the even element
for i in list:
  if i%2==0:
    print(i,end=" ")
```

```python
#Compreheogion list
a=[1,4,5,2,6,7]
l=[i for i in a if i%2==0]
print(l)
```
